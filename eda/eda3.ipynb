## Tips dataset EDA

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("tips.csv")
df.head()
```

```python
plt.scatter("total_bill","tip", data = df, label = 'tips', color="green")
plt.show()
```

```python
plt.style.use("dark_background")
```

```python
df.info()
```

```python
df.day.value_counts().index
```

```python
plt.bar(x =df.day.value_counts().index , height = df.day.value_counts().values)
plt.show()
```

```python
no1 = df["total_bill"][:31]
no2 = df["tip"][51:78]

```

```python
plt.figure(figsize=(8,4))
plt.subplot(3,3,1)
plt.bar(no1,no2,label= "1", color='red')
plt.title("BILLS")
plt.xlabel("Total bill")
plt.ylabel("Total tip") 
plt.legend()

plt.subplot(3,3,3)
plt.pie(no1, color='red')
plt.title("BILLS")
plt.xlabel("Total bill")
plt.ylabel("Total tip") 
plt.legend()

plt.subplot(3,3,5)
plt.hist(no1)
plt.title("BILLS")
plt.xlabel("Total bill")
plt.ylabel("Total tip") 
plt.legend()

plt.subplot(3,3,7)
plt.barh(no1,no2,label= "1", color='red')
plt.title("BILLS")
plt.xlabel("Total bill")
plt.ylabel("Total tip") 
plt.legend()

plt.subplot(3,3,9)
plt.scatter(no1,no2, color='red')
plt.title("BILLS")
plt.xlabel("Total bill")
plt.ylabel("Total tip") 
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(10,10))

# 1 - Top Left
plt.subplot(3,3,1)
plt.bar(no1, no2)
plt.title("Bar Chart")
plt.xlabel("Total bill")
plt.ylabel("Total tip")
plt.legend()

# 3 - Top Right
plt.subplot(3,3,3)
plt.pie(no1, colors=['red']*len(no1), autopct='%1.1f%%')
plt.title("Pie Chart")

# 5 - Center
plt.subplot(3,3,5)
plt.hist(no1)
plt.title("Histogram")
plt.xlabel("Total bill")
plt.ylabel("Frequency")

# 7 - Bottom Left
plt.subplot(3,3,7)
plt.barh(no1, no2, label="Barh", color='red')
plt.title("Horizontal Bar")
plt.xlabel("Total tip")
plt.ylabel("Total bill")
plt.legend()

# 9 - Bottom Right
plt.subplot(3,3,9)
plt.scatter(no1, no2, label="Scatter", color='red')
plt.title("Scatter Plot")
plt.xlabel("Total bill")
plt.ylabel("Total tip")
plt.legend()

plt.tight_layout()
plt.show()
```
